```python
import pandas as pd, numpy as np
from datetime import datetime, timedelta

# 10,000 rows messy data create avutundi
np.random.seed(42)
df = pd.DataFrame({
    'Order_ID': [f'ORD-{i}' for i in range(10000)],
    'Order_Date': [str(datetime(2023,1,1)+timedelta(days=np.random.randint(0,500))) for _ in range(10000)],
    'Product': np.random.choice(['Laptop','mobile','SHIRT','shoes'], 10000),
    'Quantity': np.random.choice([1,2,3,None], 10000),
    'Sale_Price': np.random.choice([50000, 2000, 999999, None], 10000),
    'Cost_Price': np.random.randint(1000,40000,10000)
})
df = pd.concat([df, df.sample(500)]) # duplicates add
df.to_csv('messy_data.csv', index=False)
print("Ready! Shape:", df.shape)
df.head()
```

```python
# --- 2nd BOX - Cleaning ---

# Load
df = pd.read_csv('messy_data.csv')

# A) Missing values entha unnayo chudu
print("Missing values:")
print(df.isnull().sum())

# B) Missing ni fill chey
df['Quantity'].fillna(1, inplace=True)
df['Sale_Price'].fillna(30000, inplace=True)

# C) Duplicates teesey
df.drop_duplicates(inplace=True)

# D) Data type fix
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Product'] = df['Product'].str.title()  # mobile -> Mobile
df['Quantity'] = df['Quantity'].astype(int)

# E) Outlier teesey (999999)
df = df[df['Sale_Price'] < 100000]

print("\nAfter cleaning:", df.shape)
df.head()
```

```python
# --- 3rd BOX - Feature Engineering ---

df['Month'] = df['Order_Date'].dt.month
df['Year'] = df['Order_Date'].dt.year
df['Profit'] = df['Sale_Price'] - df['Cost_Price']
df['Profit_Margin'] = (df['Profit'] / df['Sale_Price'] * 100).round(2)

print("New columns added!")
df.head()
```

```python
# --- 4th BOX - Export ---
df.to_csv('clean_dataset.csv', index=False)
print("Done! clean_dataset.csv ready - Shape:", df.shape)
df.head()
```
